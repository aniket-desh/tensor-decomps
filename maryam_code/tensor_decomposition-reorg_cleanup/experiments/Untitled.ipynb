```python
import numpy as np

# Parameters
rho = 1.0         # Density
T = 1.0  # Temperature
u_1d = np.array([0])        # Bulk velocity in 1D
# u_2d = np.array([0, 0])     # Bulk velocity in 2D
# u_3d = np.array([0, 0, 0])  # Bulk velocity in 3D # Bulk velocity 
d_v = len(u_1d)      # Dimensionality of velocity space
num_samples = 10


# Sample velocities

samples = np.random.normal(u_1d, np.sqrt(T), (num_samples, d_v))

def maxwell_pdf(v):
    return rho / ((2 * np.pi * T)**(d_v/2)) * np.exp(-((v - u_1d[0])**2) / (2 * T))

pdf_values = maxwell_pdf(samples[:, 0])  # Evaluate for all sampled velocities in 1D
print(pdf_values)
```

```python
import numpy as np
from sklearn.cluster import KMeans

def gaussian_basis(x, centers, sigmas):
    """
    Evaluate Gaussian basis functions at x.
    
    Parameters:
    x: array-like, shape (n_samples,)
        Points at which to evaluate the basis functions.
    centers: array-like, shape (n_centers,)
        Centers of the Gaussian basis functions.
    sigmas: array-like, shape (n_centers,)
        Standard deviations (bandwidths) for each basis.
        
    Returns:
    basis: array, shape (n_samples, n_centers)
    """
    basis = np.zeros((len(x), len(centers)))
    for i, (c, sigma) in enumerate(zip(centers, sigmas)):
        basis[:, i] = np.exp(-0.5 * ((x - c) / sigma)**2)
    return basis

# Example: Data-adaptive centers
x_data = np.random.rand(1000)  # Input data
N_centers = 10

# Place centers via k-means clustering
kmeans = KMeans(n_clusters=N_centers)
kmeans.fit(x_data.reshape(-1, 1))
centers = kmeans.cluster_centers_.flatten()

# Set bandwidths based on cluster variance
sigmas = np.array([np.std(x_data[kmeans.labels_ == i]) for i in range(N_centers)])

# Evaluate basis
x_eval = np.linspace(0, 1, 1000)
basis = gaussian_basis(x_eval, centers, sigmas)

```

```python
import numpy as np

def gaussian_basis(x, centers, sigma):
    # x: array of points where to evaluate
    # centers: array of center locations
    # sigma: scalar or array of bandwidths
    x = np.asarray(x)
    centers = np.asarray(centers)
    if np.isscalar(sigma):
        sigma = np.full_like(centers, sigma)
    basis = np.exp(-0.5 * ((x[:, None] - centers[None, :]) / sigma[None, :]) ** 2)
    return basis

# Example usage:
a, b = 0, 1  # domain
N = 10       # number of basis functions
centers = np.linspace(a, b, N)  # Equally spaced centers
sigma = (b - a) / (N - 1)       # Reasonable width

x_eval = np.linspace(a, b, 100)
basis = gaussian_basis(x_eval, centers, sigma)
print(basis.shape)

```
